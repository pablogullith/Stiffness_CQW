# 1. Abstract

<div style="
    text-align: justify;
    font-size: 16px;
    line-height: 1.6;
    text-indent: 2em;
">

We investigate boundary-induced stiffness in discrete-time quantum walks
confined to a finite lattice by unitary reflecting boundaries implemented
through coin-flip operations. We define this stiffness as the resistance of the probability distribution to spreading and redistribution over the accessible lattice. We show that confinement modifies the interference pattern of the walk from an initially ballistic, bimodal profile to a long-time coherent confined regime. In this regime, the initial sawtooth-like oscillations of the standard deviation, characterized by a finite-size oscillation period, give way at long times to smaller coherent oscillations around the value associated with a uniform distribution over the confined lattice. We also show that the normalized Kullback-Leibler distance from the uniform distribution reaches a size-dependent plateau when averaged over successive oscillation periods. Together, these results show that the long-time redistribution induced by unitary dynamics retains a strong memory of the confinement length. This study provides a framework to quantify stiffness in the crossover from ballistic propagation to finite-size coherent dynamics.

</div>

# 3. Classical Random Walk With Reflecting Boundaries

Before introducing the confined quantum walk, it is useful to recall the classical diffusion with reflecting boundary conditions. This provides the reference behavior against which the quantum dynamics will be compared. The derivation of the diffusion equation from the continuum limit of an unbiased random walk, as well as the solution of the corresponding boundary-value problem, is standard and can be found in classical treatments of diffusion.

**References:**

[The Mathematics of Diffusion](https://books.google.com.br/books?id=eHANhZwVouYC)

We consider a one-dimensional probability density \(p(x,t)\) confined to the interval \([-1,1]\). In the diffusion limit, the dynamics is governed by

\begin{equation}
\partial_t p(x,t)=D\,\partial_x^2 p(x,t),
\tag{1}
\end{equation}

where $D$ is the diffusion coefficient. Reflecting boundaries impose a vanishing probability current at the edges. Since $(J(x,t)=-D\,\partial_x p(x,t))$, this condition is equivalent to the Neumann boundary conditions $\partial_x p(\pm1,t)=0$. For an initially localized walker, $p(x,0)=\delta(x)$, the solution is even under $x\to -x$. Therefore, only the even Neumann modes contribute, and the probability density can be written as

\begin{equation}
p(x,t)=\frac12+\sum_{n=1}^{\infty}\cos(n\pi x)\,
e^{-Dn^2\pi^2t}.
\tag{2}
\end{equation}

The zero mode gives the stationary distribution, while all nonuniform modes decay exponentially. Hence, $p(x,t\to\infty)\to 1/2$ showing that the classical confined process relaxes to the uniform distribution on \([-1,1]\). This behavior is the classical signature of confinement by reflecting boundaries: large excursions are suppressed, all nonuniform spatial modes decay, and only the stationary mode survives.

The spreading of the classical probability density is characterized by the standard deviation

\begin{equation}
\sigma(t)=\sqrt{\langle x^2\rangle(t)-\langle x\rangle^2(t)}.
\tag{3}
\end{equation}

For the initial condition considered here, $\langle x\rangle(t)=0$. Using Eq. (2), one obtains

\begin{equation}
\langle x^2\rangle(t)=\frac{1}{3}+\frac{4}{\pi^2}\sum_{n=1}^{\infty}\frac{(-1)^n}{n^2}e^{-Dn^2\pi^2 t}.
\tag{4}
\end{equation}

In the long-time limit, all decaying modes vanish and $\sigma(t\to\infty)\to 1/\sqrt{3}$, which is the standard deviation of a uniform distribution on \([-1,1]\). The Figure below illustrates the classical probability distribution and the dispersion over time.


```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Math functions
def p(x, t, n_terms=10):
    result = 0.5
    for n in range(1, n_terms + 1):
        result += np.cos(n * np.pi * x) * np.exp(-n**2 * np.pi**2 * t / 2)
    return result

def sigma(t, n_terms=10):
    sum_terms = 0
    for n in range(1, n_terms + 1):
        sum_terms += ((-1)**n / n**2) * np.exp(-n**2 * np.pi**2 * t / 2)
    return np.sqrt(1/3 + (4 / np.pi**2) * sum_terms)

# 2. Generate data
x = np.linspace(-1, 1, 200)
t_array = np.linspace(0, 1, 200)
times = [0.05, 0.1, 0.25, 0.5, 1.0]

# 3. Create side-by-side subplots (1 row, 2 columns)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# First Subplot: p(x,t)
for t in times:
    ax1.plot(x, p(x, t), label=f"t={t}")

ax1.set(xlim=(-1, 1), ylim=(0, 2), xlabel='$x$', ylabel='$p(x,t)$')
ax1.set_title('Probability Density')
ax1.legend(loc='upper right')

# Second Subplot: sigma(t)
ax2.plot(t_array, sigma(t_array), color='black', label=r'$\sigma(t)$')
ax2.axhline(y=1/np.sqrt(3), color='red', linestyle='--', label='Asymptote')

ax2.set(xlim=(0, 1), ylim=(0, 1), xlabel='$t$', ylabel=r'$\sigma(t)$')
ax2.set_title('Dispersion')
ax2.legend(loc='lower right')

# 4. Clean up spacing and display
plt.tight_layout()
plt.show()
```

<div style="

In this section, we present the theoretical framework used to analyze the dynamics and stiffness of quantum walks in the presence of reflecting boundaries.

## <font color=red> Discrete-time Quantum Walks

Quantum walks describe the coherent propagation of a quantum particle on a discrete lattice, where interference effects lead to transport properties markedly different from their classical counterparts. Mathematically, their quantum state belongs to a Hilbert space that combines internal (coin) and external (position) degrees of freedom.

The Hilbert space of the discrete-time quantum walk is given by $\mathcal{H} = \mathcal{H}_c \otimes \mathcal{H}_p$, where $\mathcal{H}_c$ denotes the coin space and $\mathcal{H}_p$ the position space. The position space is spanned by the basis ${ |j\rangle ,:, j \in \mathbb{Z} }$, while the coin space by the two-dimensional basis ${ |L\rangle, |R\rangle }$. A general initial state can be written as

\begin{equation}
\lvert\Psi(0)\rangle = \sum_{j=-\infty}^{\infty}\left[\psi_L(j,0),\lvert L,j\rangle+\psi_R(j,0),\lvert R,j\rangle\right],
\tag{5}
\end{equation}

where $\psi_L(j,0)$ and $\psi_R(j,0)$ are complex initial amplitudes satisfying the normalization condition $\sum_j|\psi_L(j,0)|^2+|\psi_R(j,0)|^2=1$.

The time evolution of the discrete-time quantum walk is governed by the repeated application of a unitary operator $U$, resulting in a evolved state given by

\begin{equation}
\lvert\Psi(t)\rangle = U^t \lvert\Psi(0)\rangle.
\tag{6}
\end{equation}

The evolution operator is defined as
\begin{equation}
U = S \,(C \otimes I_p),
\tag{7}
\end{equation}

where $C$ is the coin operator acting on $\mathcal{H}_c$, $I_p$ is the identity operator on the position space, and $S$ is the conditional shift operator. We consider the standard Hadamard coin

\begin{equation}
H = \frac{1}{\sqrt{2}}
\begin{pmatrix}
1 & 1 \\
1 & -1
\end{pmatrix}
\tag{8}
\end{equation}

to our coin operator, i.e., $C=H$. Furthermore, the conditional shift operator is defined as

\begin{equation}
S = \sum_j
\left(
\lvert L\rangle\langle L\rvert \otimes \lvert j-1\rangle\langle j\rvert
+
\lvert R\rangle\langle R\rvert \otimes \lvert j+1\rangle\langle j\rvert
\right)
\tag{9}
\end{equation}

so that the walker moves to the left when the coin state is $\lvert L\rangle$ and to the right when the coin state is $\lvert R\rangle$.

## <font color=red> Reflecting Boundaries via NOT gates

We now introduce reflecting boundary conditions implemented through NOT operations acting on the coin degree of freedom.

In the bulk of the lattice, i.e., for sites satisfying $|j| < a$, the dynamics remains unchanged and the conditional shift operator acts as

\begin{align}
S\lvert L,j\rangle = \lvert L,j-1\rangle,
\tag{10} \\
S\lvert R,j\rangle = \lvert R,j+1\rangle.
\tag{11}
\end{align}

At the boundaries, however, the walker is reflected by applying a NOT operation (Pauli-$X$) on the coin state. Specifically, at the left boundary ($j=-a$) and right boundary ($j=a$), we impose

\begin{align}
S\lvert L,-a\rangle = \lvert R,-a\rangle,
\tag{12} \\
S\lvert R,a\rangle = \lvert L,a\rangle.
\tag{13}
\end{align}

Thus, whenever the walker attempts to leave the bounded lattice, its direction is reversed by flipping the coin state, effectively confining the dynamics to the finite interval $[-a,a]$.

These boundary conditions preserve the unitarity of the evolution operator while breaking translational invariance, thereby modifying the interference patterns of the walk.

## <font color=red> Initial Localized State

We first consider an initial state localized at the origin. The most general normalized coin state (qubit) can be written in the Bloch-sphere parametrization as

\begin{equation}
\lvert\chi\rangle
=
\cos\left(\frac{\alpha}{2}\right)\lvert L\rangle
+
e^{i\beta}
\sin\left(\frac{\alpha}{2}\right)\lvert R\rangle
\tag{14}
\end{equation}

where $\alpha\in[0,\pi]$ and $\beta\in[0,2\pi)$ are the polar and azimuthal angles, respectively. The corresponding initial state of the walker is

\begin{align}
\lvert\Psi(0)\rangle = \lvert 0\rangle \otimes \lvert\chi\rangle
= \cos\left(\frac{\alpha}{2}\right)\lvert 0,L\rangle + e^{i\beta}\sin\left(\frac{\alpha}{2}\right)\lvert 0,R\rangle.
\tag{15}
\end{align}

The choice of the initial coin state has a direct impact on the subsequent dynamics. For the Hadamard walk, the states $(\alpha,\beta)=(0,0)$ and $(\alpha,\beta)=(\pi,0)$, corresponding to $\lvert 0,L\rangle$ and $\lvert 0,R\rangle$, respectively, produce probability distributions biased in opposite directions. In contrast, the relative phase of the choice $(\alpha,\beta)=(\pi/2,\pi/2)$ removes the directional bias induced by the coin and yields a symmetric position distribution, $P(j,t)=P(-j,t)$, when the walker starts at the origin and the boundaries are placed symmetrically.

# 5. Results and Discussion

<div style="
    text-align: justify;
    font-size: 16px;
    line-height: 1.6;
    text-indent: 2em;
">

In a confined quantum walk, the widht of the probability distribution exhibits a quasi-periodic oscilatory dynamics generated by the repeated reflection of the balistic fronts at the NOT-gate boundaries. Since the finite unitary dynamics is not strictly periodic in general, we define an initial effective period rather than an exact recurrence time. A natural estimate for this time scale follows from the maximum group velocity of the Hadamard walk, $v_{max}=1/\sqrt{2}$, which sets the fastest propagation speed of the dominant ballistic components.

**References:**

[Group velocity of discrete-time quantum walks](https://doi.org/10.1103/PhysRevA.79.052317)

For a confinement region $j\in[-a,a]$, these fronts travel a distance of order $2a$ before returning to the central region. This gives the lower-bound time scale $T_0^*=2a/v_{\max}^*=2\sqrt{2}\,a$.

The actual period $T$ observed in the finite walk is generally larger than this estimate, $T\gtrsim T_0$, since the probability distribution contains components with different velocities and undergoes coherent interference after reflection. Operationally, in the numerical calculations, we define $T$ as the time of the first local minimum of $\sigma(t)$ after the initial expansion. This prescription avoids assuming exact periodicity and characterizes the first dominant expansion--contraction cycle of the confined distribution.

Figure bellow illustrates this cycle induced by the reflecting NOT-gate boundaries and its dynamical behavior over longer times. We also show a simulation of the walker's dynamics.


</div>

```python
import numpy as np
import matplotlib.pyplot as plt

def simulate_snapshots(N, time_steps_to_capture):
    a = (N - 1) // 2
    psi = np.zeros((2, N), dtype=complex)
    psi[0, a] = 1 / np.sqrt(2)
    psi[1, a] = 1j / np.sqrt(2)
    H = (1 / np.sqrt(2)) * np.array([[1, 1], [1, -1]])

    positions = np.arange(-a, a + 1)
    snapshots = {}

    max_t = max(time_steps_to_capture) + 1

    for t in range(max_t):
        if t in time_steps_to_capture:
            P = np.abs(psi[0])**2 + np.abs(psi[1])**2
            snapshots[t] = P.copy()

        psi_coin = np.tensordot(H, psi, axes=([1], [0]))
        psi_new = np.zeros_like(psi)

        psi_new[0, 0:N-1] = psi_coin[0, 1:N]
        psi_new[1, 1:N] = psi_coin[1, 0:N-1]

        psi_new[1, 0] = psi_coin[0, 0]
        psi_new[0, N-1] = psi_coin[1, N-1]

        psi = psi_new

    return positions, snapshots

N = 101
T = 159
t_snaps = [int(T/4), int(T/2), int(3*T/4)]

positions, snapshots = simulate_snapshots(N, t_snaps)

plt.figure(figsize=(15, 5))
titles = [r'$t \approx T/4$', r'$t \approx T/2$', r'$t \approx 3T/4$']

for i, t in enumerate(t_snaps):
    plt.subplot(1, 3, i+1)
    plt.plot(positions, snapshots[t], color='#0072B2')
    plt.title(f'{titles[i]} (t={t})')
    plt.xlabel('Position')
    if i == 0:
        plt.ylabel('Probability')
    plt.grid(True, alpha=0.3)
    # Ensure consistent y-axis for comparison
    plt.ylim(0, max([max(s) for s in snapshots.values()]) * 1.1)
    # Ensure consistent x-axis based on lattice size
    plt.xlim(-50, 50)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import Image, display

def run_simulation_animation(N, t_max):
    a = (N - 1) // 2

    psi = np.zeros((2, N), dtype=complex)
    psi[0, a] = 1 / np.sqrt(2)
    psi[1, a] = 1j / np.sqrt(2)

    H = (1 / np.sqrt(2)) * np.array([
        [1, 1],
        [1, -1]
    ])

    positions = np.arange(-a, a + 1)
    history = []

    for t in range(t_max):
        P = np.abs(psi[0])**2 + np.abs(psi[1])**2
        history.append(P)

        # Coin operation
        psi_coin = np.tensordot(H, psi, axes=([1], [0]))

        # Shift operation
        psi_new = np.zeros_like(psi)

        psi_new[0, 0:N-1] = psi_coin[0, 1:N]
        psi_new[1, 1:N] = psi_coin[1, 0:N-1]

        # Reflecting boundaries
        psi_new[1, 0] = psi_coin[0, 0]
        psi_new[0, N-1] = psi_coin[1, N-1]

        psi = psi_new

    return positions, history


# Parameters
N = 101
t_max = 160

# Run simulation
positions, history = run_simulation_animation(N, t_max)


# Create figure
fig, ax = plt.subplots(figsize=(10, 5))

line, = ax.plot([], [])

ax.set_xlim(-50, 50)
ax.set_ylim(0, 0.12)

ax.set_title('Quantum Walk Probability Distribution')
ax.set_xlabel('Position')
ax.set_ylabel('Probability')

ax.grid(True, alpha=0.3)

time_text = ax.text(
    0.02, 0.95, '',
    transform=ax.transAxes
)


def init():
    line.set_data([], [])
    time_text.set_text('')
    return line, time_text


def animate(i):
    line.set_data(positions, history[i])
    time_text.set_text(f'Time Step (t): {i}')
    return line, time_text


# Create animation
ani = animation.FuncAnimation(
    fig,
    animate,
    init_func=init,
    frames=t_max,
    interval=50,
    blit=True
)


# Save GIF
ani.save(
    'quantum_walk_N101.gif',
    writer='pillow',
    fps=20
)

plt.close(fig)


# Display GIF directly in Colab
display(Image(filename='quantum_walk_N101.gif'))
```
